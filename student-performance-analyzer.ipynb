# 🎓 Student Performance Analyzer

A beginner Python project that calculates a student's
total marks, percentage, average, grade, pass/fail status,
and highest-scoring subject.

```python
# Get student's name

name = input("Enter student name: ")

print("\nEnter marks out of 100:")

english = float(input("English: "))
mathematics = float(input("Mathematics: "))
computer = float(input("Computer: "))
physics = float(input("Physics: "))
chemistry = float(input("Chemistry: "))
```

```python
# Store subjects and their marks in a dictionary
marks = {
    "English": english,
    "Mathematics": mathematics,
    "Computer": computer,
    "Physics": physics,
    "Chemistry": chemistry
}

print("\nMarks entered successfully!")
```

```python
# Calculate total, percentage, and average
total = sum(marks.values())
total_marks = 500
percentage = (total / total_marks) * 100
average = total / len(marks)

print("Total Marks:", total)
print("Percentage:", percentage)
print("Average:", average)
```

```python
# Determine grade based on percentage
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
```

```python
# Determine pass or fail status
if percentage >= 50:
    status = "PASS"
else:
    status = "FAIL"

print("Status:", status)
```

```python
# Find the subject with the highest marks
highest_subject = max(marks, key=marks.get)
highest_marks = marks[highest_subject]

print("Highest Subject:", highest_subject)
print("Highest Marks:", highest_marks)
```
