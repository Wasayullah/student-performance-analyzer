```python
# Get student's name

name = input("Enter student name: ")

print("\nEnter marks out of 100:")

english = float(input("English: "))
mathematics = float(input("Mathematics: "))
computer = float(input("Computer: "))
physics = float(input("Physics: "))
chemistry = float(input("Chemistry: "))
# Store subjects and their marks in a dictionary
marks = {
    "English": english,
    "Mathematics": mathematics,
    "Computer": computer,
    "Physics": physics,
    "Chemistry": chemistry
}

print("\nMarks entered successfully!")


# Calculate total marks

total = sum(marks.values())

# There are 5 subjects, each having 100 marks
total_marks = 500

# Calculate percentage
percentage = (total / total_marks) * 100

# Calculate average
average = total / len(marks)

print("Total Marks:", total)
print("Percentage:", percentage)
print("Average:", average)

# Determine grade based on percentage

if percentage >= 90:
    grade = "A+"

elif percentage >= 80:
    grade = "A"

elif percentage >= 70:
    grade = "B"

elif percentage >= 60:
    grade = "C"

elif percentage >= 50:
    grade = "D"

else:
    grade = "F"

print("Grade:", grade)

# Student passes if percentage is 50 or above

if percentage >= 50:
    status = "PASS"
else:
    status = "FAIL"

print("Status:", status)

# Find the subject with the highest marks

highest_subject = max(marks, key=marks.get)
highest_marks = marks[highest_subject]

print("Highest Subject:", highest_subject)
print("Highest Marks:", highest_marks)

print("\n" + "=" * 40)
print("       STUDENT PERFORMANCE REPORT")
print("=" * 40)

print(f"\nStudent: {name}")

print("\nSubject Marks:")
print(f"English:       {marks['English']}")
print(f"Mathematics:   {marks['Mathematics']}")
print(f"Computer:      {marks['Computer']}")
print(f"Physics:       {marks['Physics']}")
print(f"Chemistry:     {marks['Chemistry']}")


print("\n" + "-" * 40)

print(f"Total:         {total} / {total_marks}")
print(f"Percentage:    {percentage:.2f}%")
print(f"Average:       {average:.2f}")


print(f"Grade:         {grade}")
print(f"Status:        {status}")
print(f"Highest:       {highest_subject} ({highest_marks})")

print("=" * 40)
```
